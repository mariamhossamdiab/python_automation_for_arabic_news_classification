```python
# run_every_30min.py
import time
import datetime
import threading
import os
import my_script
# import process_excel  # Uncomment when ready
import nlp
WATCH_FOLDER = "my_data"
CHECK_INTERVAL = 5  # seconds between folder checks
def run_scraping_every_30min():
    """Run the scraping job every 30 minutes forever."""
    while True:
        print("\n⏰ Starting new 30-minute scraping cycle...")
        start_time = datetime.datetime.now()
        print(f"🕒 Script started at: {start_time.strftime('%Y-%m-%d %H:%M:%S')}")

        try:
            my_script.main()
        except Exception as e:
            print(f"❌ Error in scraping script: {e}")

        end_time = datetime.datetime.now()
        duration = end_time - start_time
        print(f"✅ Script finished at: {end_time.strftime('%Y-%m-%d %H:%M:%S')}")
        print(f"⏱️ Duration: {duration}")

        print("💤 Sleeping for 30 minutes...\n")
        time.sleep(30 * 60)  # sleep between runs (1800 seconds)

def delete_all_files(folder_path):
    for filename in os.listdir(folder_path):
        file_path = os.path.join(folder_path, filename)
        try:
            if os.path.isfile(file_path):
                os.remove(file_path)
               # print(f"Deleted: {file_path}")
        except Exception as e:
            print(f"Error deleting {file_path}: {e}")

def watch_folder_for_csv():
    delete_all_files(WATCH_FOLDER)
    """Continuously watch a folder for new Excel files and react."""
    print(f"👀 Watching folder:{ WATCH_FOLDER}")

    # Create folder if it doesn't exist
    os.makedirs(WATCH_FOLDER, exist_ok=True)
    seen_files = set(os.listdir(WATCH_FOLDER))
    while True:
        try:
            current_files = set(os.listdir(WATCH_FOLDER))
            new_files = current_files - seen_files

            for file in new_files:
                if file.lower().endswith(".csv"):
                    file_path = os.path.join(WATCH_FOLDER, file)
                    print(f"📂 New csv file detected: {file_path}")
                    try:
                        # process_csv.main(file_path)
                        nlp.main(file_path)
                        print(f"✅ (Simulated) Processed csv file: {file}")
                    except Exception as e:
                        print(f"❌ Error processing {file}: {e}")

            seen_files = current_files
            time.sleep(CHECK_INTERVAL)

        except Exception as e:
            print(f"⚠️ Folder watch error: {e}")
            time.sleep(10)


if __name__ == "__main__":
    # Start both threads (they run forever in parallel)
    scraping_thread = threading.Thread(target=run_scraping_every_30min, daemon=True)
    folder_thread = threading.Thread(target=watch_folder_for_csv, daemon=True)

    scraping_thread.start()
    folder_thread.start()

    print("🚀 Scraper and folder watcher started!\n")

    # Keep the main thread alive so daemon threads keep running
    while True:
        time.sleep(1)
```
