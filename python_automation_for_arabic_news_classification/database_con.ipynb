```python
import mysql.connector
import pandas as pd
import numpy as np

def ccc(input_file):
    # Read CSV and replace textual NaN/NULLs with actual None
    df = pd.read_csv(f"classified_data/{input_file}")
    df.fillna("unknown", inplace=True)
    #df = df.where(pd.notnull(df), None)
    #print(df)
    conn = mysql.connector.connect(
        host="localhost",
        user="root",
        password="root",
        database="news"
    )
    cursor = conn.cursor()
    print("✅ Database connection established")

    inserted_count = 0

    for idx, row in df.iterrows():
        try:
            cursor.execute("""
                INSERT INTO classified_news (
                    Title, Link, website, Article_Text, Topic,
                    Reality, Confidence, Main_reason_for_prediction
                )
                VALUES (%s, %s, %s, %s, %s, %s, %s, %s)
            """, (
                row.get("Title"),
                row.get("Link"),
                row.get("website"),  # ✅ fixed typo here
                row.get("Article_Text"),
                row.get("Topic"),
                row.get("Reality"),
                row.get("Confidence"),
                row.get("Main_reason_for_prediction")
            ))
            inserted_count += 1
        except Exception as e:
            print(f"⚠️ Failed to insert row {idx}: {e}")
            continue

    conn.commit()
    cursor.close()
    conn.close()
    print(f"✅ Inserted {inserted_count} rows successfully")
```
