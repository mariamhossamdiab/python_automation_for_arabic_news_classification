```python
import threading
import queue
import time

# Shared queue for communication
result_queue = queue.Queue()

def task1():
    """Simulate a task that takes some time."""
    print("Task 1: Starting...")
    time.sleep(3)
    result = "Task 1 result ✅"
    print("Task 1: Done, sending result.")
    result_queue.put(result)

def task2():
    """Waits for task1's result and then processes it."""
    print("Task 2: Waiting for Task 1 to finish...")
    result = result_queue.get()  # blocks until task1 puts something
    print(f"Task 2: Got result from Task 1 → {result}")
    print("Task 2: Processing result... done ✅")

# Create threads
t1 = threading.Thread(target=task1)
t2 = threading.Thread(target=task2)

# Start both threads
t1.start()
t2.start()

# Wait for both to complete
t1.join()
t2.join()

print("All tasks complete 🎉")
```
