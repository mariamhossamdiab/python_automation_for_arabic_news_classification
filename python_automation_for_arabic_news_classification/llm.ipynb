```python
import requests

def classify_news(article_text):
    """
    Classify a news article into predefined categories using Gemma 3 (1B).
    """
    prompt = f"""
    You are a helpful assistant that classifies news articles into exactly one of these categories:
    [Politics, Sports, Technology, Business, Health, Entertainment, Science].

    Article:
    {article_text}

    Respond ONLY with the category name.
    """

    response = requests.post(
        "http://localhost:11434/api/generate",
        json={
            "model": "gemma3:1b",
            "prompt": prompt,
            "options": {
                "temperature": 0  # for consistent answers
            }
        }
    )

    # The API streams output; we’ll join all parts together
    result_text = ""
    for chunk in response.iter_lines():
        if chunk:
            data = chunk.decode("utf-8")
            if '"response"' in data:
                # Extract the text response
                part = data.split('"response":"')[-1].split('"', 1)[0]
                result_text += part

    return result_text.strip()
```

```python
import pandas as pd
df = pd.read_csv("my_data/elshrouk.csv")
# Select the first 5 articles
x = df['Article_Text'][0:5]
my_list = x.tolist()
my_list
```

```python
for i, result in enumerate(my_list):
    r = classify_news(result)
    print(f"Article {i+1}: {r}")
```
